```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use('default')
```

```python
For univarient 
1. Plotting Numerical Data (Continuous or Discrete)
 For numbers, you want to see the spread, central tendency, and shape (skewness) of the data
2. Plotting Categorical Data (Labels or Counts)
For categories, you want to see how frequently each distinct group appears
```

```python
# 1 -
price = [48000,54000,57000,49000,47000,45000]
sns.histplot(price)  #Groups values into intervals (bins) to show frequency
```

```python
sns.boxplot(price)  #Summarizes the median, quartiles, and potential outliers
```

```python
sns.kdeplot(price,fill=False)  #A smoothed, continuous version of a histogram showing probability density.
```

```python
sweets=['Rasmalai','Kalakand','Gulab Jamun','Kalakand','Rasgulla','Kalakand']
sns.countplot(sweets) #Shows the exact count of items in each category
```

```python
#for pie plot Define what those numbers represent by grouping and counting your raw data first. as it takes num cols only
x = pd.Series(sweets).value_counts()
plt.pie(x,labels=x.index, autopct='%1.1f%%')
plt.show
```
